```python
#Utilizando análise de frequência, decifre o texto cifrado abaixo. O texto em claro foi escrito em Português sem acentuação. O texto claro também usa c no lugar de ç. Submeta o texto em claro como resposta da tarefa.

TEXTO = """mjhay.
eto vmnyjrti hyqyvayi, ac iay.
nmi tsrmjht t vrmt itd naci eai,
wapt m wyhm vtooao.
a, miiyn, vmhm emiit xca hao,
fajfmoay qmgao t nasrto xca echao.
meoajhy.
jmt fmjft xcmjft xcyi,
nmi wy xca, vtjravajht
t cjywaoit mt nac oahto,
meoajht m na vtjravao nasrto,
a miiyn aivcfmoay t fanet, xca ajiyjmom
m ftnmo m havyimt vaofm an vmhm ntnajft.
a emofyoay, an dcivm ha ncyfti yhamyi.
nmi iay xca rtpa
ia ajvtjfomn nac emiimht, qcfcot a eoaiajfa.
rtpa iyjft an nyn m antvmt hm haieahyhm.
rtpa a cn etjft ha vrakmhm a,
mt naint fanet, etjft ha emofyhm.
ia an rtomi ha ajvtjfoti
etha rmwao fmjfti haiajvtjfoti,
xca m rtom hm iaemomvmt iapm, fmt-itnajfa,
m rtom ha cn waohmhayot,
eotqcjht a vtsafywt ajvtjfot.
ha fcht qyvmomt foai vtyimi:
m vaofagm ha aifmo ianeoa vtnavmjht,
m vaofagm ha xca a eoavyit vtjfyjcmo
a m vaofagm ha iao yjfaootneyht mjfai ha faonyjmo.
qmgao hm xcahm cn emiit ha hmjvm,
ht naht cnm aivmhm, ht itjrt cnm etjfa,
hm eotvcom cn ajvtjfot."""
```

```python
def frequency_analysis_simple(text):
    freq = {}
    for char in text:
        freq[char] = freq.get(char, 0) + 1
    return freq
```

```python
print(frequency_analysis_simple(TEXTO))
```

```python
# as letras que mais aparecem no texto cifrado sao as que mais aparecem em PT
cifrado = "maytcinjfhvowpxrqgedsk"
claro   = "aeiousmntdcrvjqhfzpblg"

tabela = {}
for i in range(len(cifrado)):
    tabela[cifrado[i]] = claro[i]

print(tabela)
```

```python
texto_claro = ""
for letra in TEXTO:
    if letra in tabela:
        texto_claro = texto_claro + tabela[letra]
    else:
        texto_claro = texto_claro + letra

print(texto_claro)
```

## Lógica da solução

O texto foi embaralhado com uma **cifra de substituição**: cada letra do texto original foi sempre
trocada por uma outra letra fixa. Por exemplo, todo `a` virou `m`, todo `e` virou `a`, e assim por diante.
